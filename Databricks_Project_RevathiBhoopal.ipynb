## **Import the Libraries**

```python
import pandas as pd 
import numpy as np
```

**## Read data into dataframe from csv**

```python
url = 'https://raw.githubusercontent.com/a-forty-two/DFE6/main/titanic.csv'
titanic = pd.read_csv(url)
titanic.head()
```

## Data Cleansing

**## Capitalize and remove null values from all Columns**

```python
titanic = titanic.set_index("PassengerId")
titanic = titanic.applymap(lambda x: x.capitalize() if isinstance(x, str) else x)
titanic = titanic.dropna()
titanic = titanic.drop(['Parch', 'SibSp', 'Embarked'], axis=1)
titanic
```

## **Write the dataframe into csv and store in Volumes**

```python
%sql
CREATE VOLUME IF NOT EXISTS titanic_volume
```

```python
volume_path = "/Volumes/pikapika/default/titanic_volume/titanic_cleansed_data.csv"
titanic.to_csv(volume_path, index=True)
```
